```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException
from bs4 import BeautifulSoup
import os
import time
```

```python
driver = webdriver.Chrome()
driver.get('https://www.pakwheels.com/')
items = driver.find_elements(By.CSS_SELECTOR, 'ul.browse-listing li.col-sm-2 a')
output_dir = "html_outputs"
os.makedirs(output_dir, exist_ok=True)
count = 0

for idx, category in enumerate(items, start=1):
    outer_html = category.get_attribute("outerHTML")
    file_path = os.path.join(output_dir, f"category_{idx}.html")
    
    with open(file_path, "w", encoding="utf-8") as file:
        file.write(outer_html)
    print(f"Wrote: {file_path}")
    count += 1
    if(count == 40):
        break

driver.quit()
```

```python
directory = "html_outputs"
dic = []
BASE_URL = "https://www.pakwheels.com"

for filename in os.listdir(directory):
    filepath = os.path.join(directory, filename)
    if os.path.isfile(filepath):
        with open(filepath, 'r', encoding='utf-8') as f:
            content = f.read()
            soup = BeautifulSoup(content, 'html.parser')
            
            links = soup.find_all("a", href=True)

            for tag in links:
                tempdic = {'name': tag.text.strip() , 'urls': BASE_URL+tag['href']}
                dic.append(tempdic)
print(dic)


    
```

```python
import requests  
from bs4 import BeautifulSoup
import os

os.makedirs('cars', exist_ok=True)
header = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                  '(KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36'
}

for item in dic:
    all_listings = []
    
    for page_num in range(1, 21):
        page_url = f"{item['urls']}?page={page_num}"
        try:
            response = requests.get(page_url, headers=header)
            if response.status_code == 404:
                print(f"[404] Skipping page {page_num} for {item['name']}")
                break
            response.raise_for_status()
            soup = BeautifulSoup(response.text, 'html.parser')
            listings = soup.find_all('li', class_='classified-listing')
            
            if not listings:
                print(f"No listings found on page {page_num} for {item['name']}. Stopping early.")
                break
            all_listings.extend(listings)
            print(f"Fetched page {page_num} for {item['name']} with {len(listings)} listings")
        except requests.exceptions.RequestException as e:
            print(f"Error fetching page {page_num} for {item['name']}: {e}")
            break
    # Save all listings to one file per category
    if all_listings:
        file_path = os.path.join('cars', f"{item['name']}.html")
        with open(file_path, "w", encoding="utf-8") as file:
            for listing in all_listings:
                file.write(listing.prettify())
                file.write("\n" + "="*50 + "\n")
        print(f"✅ Saved {len(all_listings)} listings for {item['name']}")
```

```python
import pandas as pd
directory = 'cars'
all_data = []

for filename in os.listdir(directory):

    filepath = os.path.join(directory, filename)
    if os.path.isfile(filepath):
        with open(filepath, 'r', encoding='utf-8') as f:
            base_name = os.path.basename(f.name)
            category = os.path.splitext(base_name)[0]
            soup = BeautifulSoup(f, 'html.parser')

            for i in range(1,21):
                listings = soup.find_all('div', class_='well search-list clearfix ad-container page-{}'.format(i))
                for listing in listings:
                    try:
                        title_tag = listing.find('a', class_='car-name ad-detail-path')
                        title = title_tag.get_text(strip=True) if title_tag else 'N/A'
                        url = title_tag['href'] if title_tag and title_tag.has_attr('href') else 'N/A'

                        price_tag = listing.find('strong', class_='generic-white fs18')
                        price = price_tag.get_text(strip=True) if price_tag else 'N/A'

                        city_tag = listing.find('ul', class_='list-unstyled search-vehicle-info fs13')
                        city = city_tag.find('li').get_text(strip=True) if city_tag else 'N/A'

                        details = listing.find('ul', class_='list-unstyled search-vehicle-info-2 fs13')
                        year = details.find_all('li')[0].get_text(strip=True) if details else 'N/A'
                        mileage = details.find_all('li')[1].get_text(strip=True) if details else 'N/A'
                        fuel = details.find_all('li')[2].get_text(strip=True) if details else 'N/A'
                        engine = details.find_all('li')[3].get_text(strip=True) if details else 'N/A'
                        transmission = details.find_all('li')[4].get_text(strip=True) if details else 'N/A'

                        all_data.append({
                            "category": category,
                            "title": title,
                            "price": price,
                            "city": city,
                            "year": year,
                            "mileage": mileage,
                            "fuel": fuel,
                            "engine": engine,
                            "transmission": transmission,
                            "url": f"https://www.pakwheels.com{url}"
                        })
                    except Exception as e:
                        print(f"Error parsing listing in file {filename}: {e}")
                print("Saved the data of {}".format(category))

df = pd.DataFrame(all_data)
df = df.sample(frac=1).reset_index(drop=True)
df.to_csv('car_listings.csv', index=False, encoding='utf-8')
print("CSV file saved as 'car_listings.csv'")

```
